```python
import matplotlib
import seaborn
import plotly
import pandas
print("Matplotlib:", matplotlib.__version__)
print("Seaborn:", seaborn.__version__)
print("Plotly:", plotly.__version__)
print("Pandas:", pandas.__version__)
```

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,15,12,18,20]
plt.plot(x, y)
plt.title("Line Chart")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = ["A", "B", "C", "D", "E"]
y = [10, 15, 12, 18, 20]
plt.bar(x, y)
plt.title("Bar Graph")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = ["A", "B", "C", "D"]
values = [30, 25, 20, 25]
plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 18, 20]
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

```python
import matplotlib.pyplot as plt
data = [10, 12, 15, 15, 18, 20, 20, 22, 25, 25]
plt.hist(data, bins=5)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
x <- c(1,2,3,4,5)
y <- c(10,15,12,18,20)

plot(x, y,
     type="o",
     col="blue",
     main="Line Graph",
     xlab="X Values",
     ylab="Y Values")
```

```python
x <- c("A","B","C","D","E")
y <- c(10,15,12,18,20)
barplot(y,
        names.arg=x,
        col="blue",
        main="Bar Graph",
        xlab="Categories",
        ylab="Values")
```

```python
x <- c(30,25,20,25)
labels <- c("A","B","C","D")
pie(x,
    labels=labels,
    col=c("red","blue","green","yellow"),
    main="Pie Chart")
```

```python
x <- c(1,2,3,4,5)
y <- c(10,15,12,18,20)
plot(x, y,
     type="p",
     col="red",
     main="Scatter Plot",
     xlab="X Values",
     ylab="Y Values")
```
